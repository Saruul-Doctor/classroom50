```python
X = df.drop(columns=['Target'])
X = X.replace(['?', 'Unknown', 'Five', 'Na'], np.nan)
```

```python
print("Missing values per colum before transformation:")
print(X.isnull().sum)
```

```python
df['Target'].value_counts()
```

```python
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors='coerce')

X = X.fillna(X.mean())
print(f"Total remaining missing values is features: {X.isnull().sum().sum()}")
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca_full = PCA()
pca_full.fit(X_scaled)

cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

n_components_995 = np.argmax(cumulative_variance >= 0.995) + 1

print(f"Minimum numbr of features required to retain 99.5% variance: {n_components_995}")
```

```python
pca_final = PCA(n_components_995)
X_pca = pca_final.fit_transform(X_scaled)

print(f"New shape of the dataset after dimensionality reduction: {X_pca.shape}")
```

```python
plt
```

```python
pca_df = pd.DataFrame(data=X_pca, columns=[f'PC{i+1}' for i in range(n_components_995)])
pca_df['Target'] = df['Target'].values

#Project records on a 2D scatter grid using PC1 and PC2
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='Target',
    alpha=0.7
)
plt.title('Practical Test: Projection of Student Targets on PC1 and PC2')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.legend(title='Student status')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Curse of Dimensionality 
The "Curse of Dimensionality" happens when you add too many columns (features) to your data. As the number of features increases, the data space grows so fast that the data points become very spread out and far apart. Because of this, traditional ways to measure distance, like Euclidean distance, do not work well anymore, and distance-based algorithms star to make mictakes.

**Using our 36-attribute dataset directly creates the following risks:**

1. **Overfitting:** The model learns random noise in the data instead of the real patterns.
2. **Computational detay:** Many correlated columns slow down matrix calculations significantly.

We successfully solved this problem by using a PCA model. By reducing the data to 33 components, we kept 99.5% of the total information while safely avoiding the risks of the "curse of dimensionality.

# Practical testing Discussion
The practical testing scatter plot shows the data projected onto the thPC1 and PC2 axes. Even though we reduced the dimensions from 36 to 33, the main patterns in the data remain the same. The data points for "Graduate" and "Dropout" students are clearly separated and form nice clusters. This proves that the smaller data space completely keeps its predictive power.

# Conclusions and Findings 
* **Pre-processing cleanup:** The strategy of turning non-numeric error characters (noise) into NaN and filling them with the average value was very successful. This solved the problemo of category inflation.

* **Dimensionality reduction:** We optimized the high correlation between semester grades (multicollinearity). We successfully moved the data into a 33-dimensional space while keeping 99.5% of the original information (variance).

* **Practical proof:** The visualization on the PC1 and PC2 axes shows that our data did not lose important classification information, even after reducing the dimensions.
